### 내 드라이브 파일 목록 확인
마운트가 완료되면 아래 코드를 통해 '내 드라이브(My Drive)'에 있는 폴더와 파일 목록을 확인할 수 있습니다.

```python
import os

# 구글 드라이브 내 'My Drive' 경로
drive_path = '/content/drive/MyDrive'

if os.path.exists(drive_path):
    print("구글 드라이브 연결 성공!")
    print("내 드라이브 파일 목록:")
    display(os.listdir(drive_path)[:10]) # 상위 10개 파일/폴더만 표시
else:
    print("드라이브를 찾을 수 없습니다. 마운트 상태를 확인해 주세요.")
```
